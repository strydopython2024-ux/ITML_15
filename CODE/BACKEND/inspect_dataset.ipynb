```python
import pandas as pd
import os

# Dataset path
DATASET_PATH = "../DATA_SET/Crop recommendation dataset.csv"

# Load dataset
df = pd.read_csv(DATASET_PATH)

print("=" * 70)
print("SMART FARMING - DATASET INSPECTION")
print("=" * 70)

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 10 Rows:")
display(df.head(10))

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nStatistical Summary:")
display(df.describe(include="all").T)
```

```python
print("=" * 70)
print("UNIQUE VALUES FOR EACH COLUMN")
print("=" * 70)

for column in df.columns:
    print(f"\n{column}")
    print("-" * 40)
    print("Unique values:", df[column].nunique())

    if df[column].nunique() <= 30:
        print(df[column].unique())
```

```python
print("=" * 70)
print("SOIL TYPE CHECK")
print("=" * 70)

soil_columns = [
    "soil_type",
    "Soil_Type",
    "soil type",
    "Soil Type",
    "soil"
]

found_soil = []

for column in soil_columns:
    if column in df.columns:
        found_soil.append(column)

if found_soil:

    print("Soil Type column found:")
    print(found_soil)

    for column in found_soil:
        print(f"\n{column} values:")
        print(df[column].value_counts())

else:

    print("NO SOIL TYPE COLUMN FOUND")
    print("\nAvailable columns are:")
    print(df.columns.tolist())
```

```python
print("=" * 70)
print("CROP CLASSES")
print("=" * 70)

possible_crop_columns = [
    "crop",
    "Crop",
    "label",
    "Label",
    "crop_name",
    "Crop_Name"
]

for column in possible_crop_columns:

    if column in df.columns:

        print("Crop column:", column)
        print("\nNumber of classes:", df[column].nunique())

        print("\nCrop distribution:")
        display(df[column].value_counts())

        break
    
```

```python
print("=" * 70)
print("CATEGORICAL COLUMN VALUES")
print("=" * 70)

categorical_columns = [
    "CROPS",
    "TYPE_OF_CROP",
    "SOIL",
    "SEASON",
    "SOWN",
    "HARVESTED",
    "WATER_SOURCE"
]

for column in categorical_columns:
    print(f"\n{'=' * 50}")
    print(f"{column}")
    print(f"{'=' * 50}")
    print("Unique values:", df[column].nunique())
    print(df[column].dropna().unique()[:100])
```

```python
print("=" * 70)
print("NUMERICAL DATA SUMMARY")
print("=" * 70)

numeric_columns = [
    "SOIL_PH",
    "SOIL_PH_HIGH",
    "CROPDURATION",
    "CROPDURATION_MAX",
    "TEMP",
    "MAX_TEMP",
    "WATERREQUIRED",
    "WATERREQUIRED_MAX",
    "RELATIVE_HUMIDITY",
    "RELATIVE_HUMIDITY_MAX",
    "N",
    "N_MAX",
    "P",
    "P_MAX",
    "K",
    "K_MAX"
]

display(df[numeric_columns].describe().T)
```

```python
print("=" * 70)
print("CROP DISTRIBUTION")
print("=" * 70)

print("Number of crop classes:", df["CROPS"].nunique())

display(
    df["CROPS"]
    .value_counts()
    .to_frame("Count")
)
```

```python
print("SOIL TYPES IN THE DATASET")
print("=" * 50)

soil_types = df["SOIL"].dropna().unique()

print("Number of soil types:", len(soil_types))
print("\nSoil types:")

for i, soil in enumerate(soil_types, 1):
    print(f"{i}. {soil}")

print("SOIL TYPE DISTRIBUTION")
print("=" * 50)

print(df["SOIL"].value_counts())    
```

```python


# Clean SOIL values
df["SOIL_CLEAN"] = (
    df["SOIL"]
    .astype(str)
    .str.replace("\u00a0", " ", regex=False)  # Remove non-breaking spaces
    .str.strip()                              # Remove leading/trailing spaces
    .str.lower()                              # Convert to lowercase
    .str.replace(r"\s+", " ", regex=True)     # Remove duplicate spaces
)

print("RAW SOIL TYPES :", df["SOIL"].nunique())
print("CLEAN SOIL TYPES:", df["SOIL_CLEAN"].nunique())

print("\nCLEAN SOIL TYPES")
print("=" * 50)

for i, soil in enumerate(sorted(df["SOIL_CLEAN"].unique()), 1):
    print(f"{i}. {soil}")
```

```python
print("CLEAN SOIL TYPE DISTRIBUTION")
print("=" * 50)

print(df["SOIL_CLEAN"].value_counts())
```
